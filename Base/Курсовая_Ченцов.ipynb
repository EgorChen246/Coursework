Бизнес-цель проекта: создать инструмент, который по открытым и разнообразным данным будет сигнализировать о изменении тренда акций и искать повторяющиеся паттерны для улучшения предсказания аналитикам и инвесторам.
 
Российский фондовый рынок имеет высокую волатильность, поэтому инвесторам важно знать будущее движение акции. Если такое движение удастся распознать в самом начале, то получится скорректировать и пересмотреть свой инвестиционный портфель. 
 
Критерии достижения цели: инструмент обученный на исторических данных будет устойчиво определять периоды повышенной волатильности от спокойной. При этом адаптируемый под разные типы данных и воспроизводимый на новых. В момент обучения не будет использоваться информация из будущего, чтобы предотвратить переобучение модели
 
Исследовательская задача: Проверить совместимость различных котировок, факторов и данных для классификации ожидаемого направления на горизонте 5 дней и сравнить с моделью, построенную на исходных котировках.
 
Гипотеза: Улучшается ли качество предсказания тренда акции при добавлении технических, макроэкономических и текстовых признаков относительно базовой модели обученной на одних котировках.
 
Дополнительная гипотеза: Получиться ли определить при правильном предсказании, что сильнее всего повлияло на направление движения цены акции
 
Границы исследования: Период анализа взят период с 2022 года, поскольку после санкционных ограничений и сокращения иностранного финансирования, рыночные условия на бирже сильно изменились, и более ранние данные отражают не правдивую и не реалистичную в нынешних условиях среду
 
Объект исследования: котировка акций и рыночные тренды на Московской бирже
 
Предмет исследования: статистические и технические зависимости между историческими котировками, макроэкономическими факторами, новостными статьями и будущим направлением движения котировок 
 
Проверка гипотезы: Для проверки будут реализованы и обучены три модели - Logistic Regression, Random Forest, Gradient Boosting
